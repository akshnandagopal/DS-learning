```python
import pandas as pd
import numpy as np
from sklearn.cluster import AgglomerativeClustering
```

```python
#1 Load the file “zoo.csv” and look at the info and the first five rows. 
# The first column denotes the animal name and 
# the last one specifies a high-level class for the corresponding animal.

df = pd.read_csv('Datasets/additional_resources_10_6qt_ynxafhm/zoo.csv')
print(df.shape)
df.head(5)
```

```python
df.info()
```

```python
df['class_type'].value_counts()
```

```python
print(df.duplicated().sum())

df.isnull().sum()
```

```python
#2 Find out the unique number of high-level classes

df['class_type'].nunique()
```

```python
#3 Use the 16-intermediate feature and perform an agglomerative clustering.
# [ Hint: Refer to the agglomerative clustering module in scikit learn and 
# set the number of clusters appropriately]

X = df.drop(['animal_name', 'class_type'], axis=1)   # features

cluster_model = AgglomerativeClustering(n_clusters=7)
predicted = cluster_model.fit_predict(X)
```

```python
predicted
```

```python
actual_class = np.array(df['class_type'])   #actual
actual_class
```

```python
cluster_model.linkage
```

```python
X
```

```python
#4 Compute the mean squared error by comparing the actual class and predicted high-level class.

from sklearn.metrics import mean_squared_error
from sklearn.metrics import silhouette_score

print(f"Mean Squared Error: {mean_squared_error(actual_class,predicted): .2f}")  # mse is not commonly used for clustering
                                                                                # data labels in actual_class: 1-7, in predicted: 0-6

print(f"Silhouette Score: {silhouette_score(X,predicted): .2f}")
```
