```python
# encode book codes

ratings['isbn_encoded'] = ratings['isbn'].astype('category').cat.codes
```

```python
print(ratings['isbn'])
ratings['isbn_encoded']
```

```python
from sklearn.neighbors import NearestNeighbors

X = ratings[['user_id','isbn_encoded','rating']]
knn = NearestNeighbors(metric='cosine', algorithm='brute')
knn.fit(X.values)
```

```python
# example for user 2

target_user = X[X['user_id'] == 2].values
distances, indices = knn.kneighbors(target_user, n_neighbors=3)
```

```python
print(distances)
indices
```

```python
# neighbors and recommend books

neighbor_data = X.iloc[indices.flatten()]
recommended_books = neighbor_data['isbn_encoded'].unique()
```

```python
# decode book codes

original_book_code = ratings[ratings['isbn_encoded'].isin(recommended_books)]['isbn'].unique()
```
