```python
#1 Load the kinematics dataset as measured on mobile sensors from the file “run_or_walk.csv”.
#  List out the columns in the dataset.

df = pd.read_csv('Datasets/additional_resources_9_ox6_o5534js/run_or_walk.csv')
df
```

```python
#2 Let the target variable ‘y’ be the activity and assign all the columns after it to ‘x’. 

X = df.drop(['date','time','username','activity'], axis=1)
y = df['activity']
```

```python
X, y
```

```python
#3 Using Scikit-learn fit a Gaussian Naive Bayes model and observe the accuracy. 
# Generate a classification report using scikit learn. 

from sklearn.naive_bayes import GaussianNB

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

gnb = GaussianNB()
gnb.fit(X_train, y_train)

ypred = gnb.predict(X_test)

# accuracy score
print('Accuracy:', accuracy_score(y_test, ypred))

# classification report
print('\nClassification report:\n',classification_report(y_true=y_test, y_pred=ypred))
```

```python
#4 Repeat the model once using only the acceleration values as predictors and then 
# using only the gyro values as predictors. 
# Comment on the difference in accuracy between both the models

# Acceleration as predictos

X_acc = df[['acceleration_x','acceleration_y','acceleration_z']]
X_train_acc, X_test_acc, y_train_acc, y_test_acc = train_test_split(X_acc, y, test_size=0.2)

gnb_acc = GaussianNB()
gnb_acc.fit(X_train_acc, y_train_acc)

ypred_acc = gnb_acc.predict(X_test_acc)

# accuracy score
print('Accuracy:', accuracy_score(y_test_acc, ypred_acc))
```

```python
# Gyros as predictos

X_gyr = df[['gyro_x','gyro_y','gyro_z']]

X_train_gyr, X_test_gyr, y_train_gyr, y_test_gyr = train_test_split(X_gyr, y, test_size=0.2)

gnb_gyr = GaussianNB()
gnb_gyr.fit(X_train_gyr, y_train_gyr)

ypred_gyr = gnb_gyr.predict(X_test_gyr)

# accuracy score
print('Accuracy:', accuracy_score(y_test_gyr, ypred_gyr))
```

```python
print('Model with Acceleration as predictors shows more accuracy of 95%')
```

```python
# exploring confusion matrix

cm = confusion_matrix(y_test,ypred)
cm_acc = confusion_matrix(y_test_acc,ypred_acc)

print(cm.sum() - cm.trace())
print(cm_acc.sum() - cm_acc.trace())
```
