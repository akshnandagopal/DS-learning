```python
#1 Load the data from “College.csv” that has attributes collected about private 
# and public colleges for a particular year. 
# We will try to predict the private/public status of the college from other attributes.


df = pd.read_csv('Datasets/additional_resources_9_ox6_o5534js/College.csv')
df
```

```python
df['Private'].value_counts()
```

```python
df.info()
```

```python
#2 Use LabelEncoder to encode the target variable into numerical form and 
# split the data such that 20% of the data is set aside for testing.

from sklearn.preprocessing import LabelEncoder

encode = LabelEncoder()
df['Private'] = encode.fit_transform(df['Private'])
df[['Private']]
```

```python
from sklearn.model_selection import train_test_split

X = df.drop('Private', axis=1)
y = df['Private']    # cannot pass a dataframe(or column vector) to y as the model expects 1d array

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
#3 Fit a linear SVM from scikit learn and observe the accuracy. 
# [Hint:Use Linear SVC]

from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

svm = LinearSVC()
svm.fit(X_train, y_train)

ypred = svm.predict(X_test)

score = accuracy_score(y_test, ypred)
print('Accuracy score:', score)
```

```python
# finding the score using scikit score module

print(svm.score(X_test,y_test))
```

```python
#4 Preprocess the data using StandardScalar and fit the same model again and observe the change in accuracy.
# [Hint: Refer to scikitlearn’s preprocessing methods]

from sklearn.preprocessing import StandardScaler

scale = StandardScaler()
svm_scaled = LinearSVC()

X_train_scaled = scale.fit_transform(X_train)
X_test_scaled = scale.transform(X_test)

svm_scaled.fit(X_train_scaled, y_train)

print('Accuracy score:', svm_scaled.score(X_test_scaled, y_test))
```

```python
X_train_scaled
```

```python
svm_scaled.get_params()
```

```python
#5 Use scikit learns grid search to select the best hyperparameter for a non-linear SVM, and 
# identify the model with the best score and its parameters. 
# [Hint: Refer to model_selection module of Scikit learn]

from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV

params = { 'C' : [0.1,1,10,100],
          'kernel' : ['rbf'],
          'gamma' : [0.001,0.01,0.1,1]
            }
gs = GridSearchCV(estimator=SVC(), param_grid = params, cv=5, scoring='accuracy', verbose=1)

gs.fit(X_train_scaled, y_train)
```

```python
print(gs.best_score_)
print(gs.best_params_)
print(gs.best_estimator_)
gs.score(X_test_scaled,y_test)
```
